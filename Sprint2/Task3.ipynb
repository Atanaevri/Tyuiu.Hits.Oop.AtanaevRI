```
using System;
using System.Collections.Generic;

// ===== ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ =====
public interface IDigitalMedia
{
    string FileFormat { get; set; }
    void Download();
}

public interface IAudioBook
{
    int DurationMinutes { get; set; }
    void Listen();
}

public interface IInteractive
{
    void LaunchInteractiveMode();
}

// 1. Базовый класс Book
public class Book
{
    private string _title;
    private string _author;
    private int _yearOfPublication;
    private string _isbn;
    private int _pages;
    private string _publisher;
    private bool _isAvailable;

    public string Title { get { return _title; } set { _title = value; } }
    public string Author { get { return _author; } set { _author = value; } }
    public int YearOfPublication { get { return _yearOfPublication; } set { _yearOfPublication = value; } }
    public string ISBN { get { return _isbn; } set { _isbn = value; } }
    public int Pages { get { return _pages; } set { _pages = value; } }
    public string Publisher { get { return _publisher; } set { _publisher = value; } }
    public bool IsAvailable { get { return _isAvailable; } set { _isAvailable = value; } }

    public Book(string title, string author, int yearOfPublication,
                string isbn = "N/A", int pages = 0,
                string publisher = "Неизвестно", bool isAvailable = true)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        ISBN = isbn;
        Pages = pages;
        Publisher = publisher;
        IsAvailable = isAvailable;
    }

    public virtual string GetInfo()
    {
        return $"Название: \"{Title}\", Автор: {Author}, Год издания: {YearOfPublication}, " +
               $"ISBN: {ISBN}, Страниц: {Pages}, Издательство: {Publisher}, Доступна: {IsAvailable}";
    }

    public string GetShortInfo() { return $"{Title} — {Author} ({YearOfPublication})"; }
    public void MarkAsBorrowed() { IsAvailable = false; }
    public void ReturnBook()
    {
        IsAvailable = true;
        Console.WriteLine($"Книга \"{Title}\" возвращена в библиотеку.");
    }

    public void ChangePublisher(string newPublisher)
    {
        Publisher = newPublisher;
        Console.WriteLine($"Издательство книги \"{Title}\" изменено на: {Publisher}");
    }

    public virtual void Read() { Console.WriteLine($"Чтение книги \"{Title}\"..."); }

    public virtual void Borrow()
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"Книга \"{Title}\" недоступна (уже выдана).");
            return;
        }
        MarkAsBorrowed();
        Console.WriteLine($"Книга \"{Title}\" выдана на чтение.");
    }
}

// 2. ПРОСТОЕ НАСЛЕДОВАНИЕ: Textbook : Book
public class Textbook : Book
{
    private string _subject;
    private int _gradeLevel;
    private bool _hasExercises;
    private string _disciplineCode;

    public string Subject { get { return _subject; } set { _subject = value; } }
    public int GradeLevel { get { return _gradeLevel; } set { _gradeLevel = value; } }
    public bool HasExercises { get { return _hasExercises; } set { _hasExercises = value; } }
    public string DisciplineCode { get { return _disciplineCode; } set { _disciplineCode = value; } }

    public Textbook(string title, string author, int year, string subject,
                    int gradeLevel = 0, bool hasExercises = true, string disciplineCode = "N/A")
        : base(title, author, year)
    {
        Subject = subject;
        GradeLevel = gradeLevel;
        HasExercises = hasExercises;
        DisciplineCode = disciplineCode;
    }

    public override void Read() { Console.WriteLine($"Чтение учебника \"{Title}\" по предмету: {Subject}."); }
    public string GetSubjectInfo() { return $"Предмет: {Subject}, Класс: {GradeLevel}, Код дисциплины: {DisciplineCode}"; }
    public void OpenExercise(int number) { Console.WriteLine($"Открыто упражнение №{number} в учебнике \"{Title}\"."); }
    public override string GetInfo() { return base.GetInfo() + $", Предмет: {Subject}, Класс: {GradeLevel}, Есть упражнения: {HasExercises}"; }
}

// 3. ПРОСТОЕ НАСЛЕДОВАНИЕ: Fiction : Book
public class Fiction : Book
{
    private string _genre;
    private string _ageRating;
    private bool _isSeries;
    private string _seriesName;

    public string Genre { get { return _genre; } set { _genre = value; } }
    public string AgeRating { get { return _ageRating; } set { _ageRating = value; } }
    public bool IsSeries { get { return _isSeries; } set { _isSeries = value; } }
    public string SeriesName { get { return _seriesName; } set { _seriesName = value; } }

    public Fiction(string title, string author, int year, string genre,
                   string ageRating = "0+", bool isSeries = false, string seriesName = "Нет")
        : base(title, author, year)
    {
        Genre = genre;
        AgeRating = ageRating;
        IsSeries = isSeries;
        SeriesName = seriesName;
    }

    public override void Borrow()
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"Художественная книга \"{Title}\" недоступна.");
            return;
        }
        MarkAsBorrowed();
        Console.WriteLine($"Выдача художественной книги \"{Title}\" в жанре: {Genre}.");
    }

    public string GetGenreInfo() { return $"Жанр: {Genre}, Возрастной рейтинг: {AgeRating}, Серия: {(IsSeries ? SeriesName : "нет")}"; }
    public void ReadChapter(int chapter) { Console.WriteLine($"Чтение главы {chapter} книги \"{Title}\"."); }
    public override string GetInfo() { return base.GetInfo() + $", Жанр: {Genre}, Рейтинг: {AgeRating}, Серия: {(IsSeries ? SeriesName : "нет")}"; }
}

// 4. ПРОСТОЕ НАСЛЕДОВАНИЕ: ScientificLiterature : Book
public class ScientificLiterature : Book
{
    private string _fieldOfScience;
    private string _doi;
    private int _citationCount;
    private bool _peerReviewed;

    public string FieldOfScience { get { return _fieldOfScience; } set { _fieldOfScience = value; } }
    public string DOI { get { return _doi; } set { _doi = value; } }
    public int CitationCount { get { return _citationCount; } set { _citationCount = value; } }
    public bool PeerReviewed { get { return _peerReviewed; } set { _peerReviewed = value; } }

    public ScientificLiterature(string title, string author, int year, string fieldOfScience,
                                string doi = "N/A", int citationCount = 0, bool peerReviewed = false)
        : base(title, author, year)
    {
        FieldOfScience = fieldOfScience;
        DOI = doi;
        CitationCount = citationCount;
        PeerReviewed = peerReviewed;
    }

    public string GetScienceInfo() { return $"Область науки: {FieldOfScience}, DOI: {DOI}, Цитирований: {CitationCount}, Рецензируемая: {PeerReviewed}"; }
    public void Cite()
    {
        CitationCount++;
        Console.WriteLine($"Работа \"{Title}\" процитирована. Всего цитирований: {CitationCount}");
    }
    public override string GetInfo() { return base.GetInfo() + $", Область науки: {FieldOfScience}, DOI: {DOI}, Цитирований: {CitationCount}, Рецензируемая: {PeerReviewed}"; }
}

// ===== СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ =====
public class IllustratedTextbook : Textbook
{
    private string _illustrator;
    private int _illustrationCount;
    private string _format;

    public string Illustrator { get { return _illustrator; } set { _illustrator = value; } }
    public int IllustrationCount { get { return _illustrationCount; } set { _illustrationCount = value; } }
    public string Format { get { return _format; } set { _format = value; } }

    public IllustratedTextbook(string title, string author, int year, string subject,
                               int gradeLevel, bool hasExercises, string disciplineCode,
                               string illustrator, int illustrationCount, string format)
        : base(title, author, year, subject, gradeLevel, hasExercises, disciplineCode)
    {
        Illustrator = illustrator;
        IllustrationCount = illustrationCount;
        Format = format;
    }

    public void ShowIllustration(int number) { Console.WriteLine($"Показ иллюстрации №{number} из учебника \"{Title}\" (иллюстратор: {Illustrator})."); }
    public string GetIllustratedInfo() { return $"Иллюстратор: {Illustrator}, Иллюстраций: {IllustrationCount}, Формат: {Format}"; }
    public override string GetInfo() { return base.GetInfo() + $", Иллюстратор: {Illustrator}, Иллюстраций: {IllustrationCount}, Формат: {Format}"; }
}

// ===== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ =====
public class MultimediaBook : Book, IDigitalMedia, IAudioBook, IInteractive
{
    private string _fileFormat;
    private int _durationMinutes;
    private bool _hasInteractiveMode;

    public string FileFormat { get { return _fileFormat; } set { _fileFormat = value; } }
    public int DurationMinutes { get { return _durationMinutes; } set { _durationMinutes = value; } }
    public bool HasInteractiveMode { get { return _hasInteractiveMode; } set { _hasInteractiveMode = value; } }

    public MultimediaBook(string title, string author, int year,
                          string isbn, int pages, string publisher, bool isAvailable,
                          string fileFormat, int durationMinutes, bool hasInteractiveMode)
        : base(title, author, year, isbn, pages, publisher, isAvailable)
    {
        FileFormat = fileFormat;
        DurationMinutes = durationMinutes;
        HasInteractiveMode = hasInteractiveMode;
    }

    public void Download() { Console.WriteLine($"Скачивание мультимедийной книги \"{Title}\" в формате {FileFormat}."); }
    public void Listen() { Console.WriteLine($"Прослушивание аудиоверсии \"{Title}\" ({DurationMinutes} мин.)."); }
    public void LaunchInteractiveMode()
    {
        if (!HasInteractiveMode)
        {
            Console.WriteLine($"У книги \"{Title}\" нет интерактивного режима.");
            return;
        }
        Console.WriteLine($"Запущен интерактивный режим книги \"{Title}\".");
    }
    public override string GetInfo() { return base.GetInfo() + $", Формат: {FileFormat}, Длительность аудио: {DurationMinutes} мин., Интерактив: {HasInteractiveMode}"; }
}

// 5. Класс Читатель (Reader)
public class Reader
{
    public string Name { get; set; }
    public Reader(string name) { Name = name; }

    public void TakeBook(Book book)
    {
        Console.WriteLine($"\n[Читатель {Name}] берет книгу \"{book.Title}\".");
        book.Borrow();
    }
}

// 6. Класс Библиотека (Library)
public class Library
{
    private List<Book> _books = new List<Book>();

    public void AddBook(Book book)
    {
        _books.Add(book);
        Console.WriteLine($"[Библиотека] Добавлена книга: \"{book.Title}\"");
    }

    public void LendBook(Reader reader, string title)
    {
        Book foundBook = null;
        foreach (var book in _books)
        {
            if (book.Title.Equals(title, StringComparison.OrdinalIgnoreCase))
            {
                foundBook = book;
                break;
            }
        }

        if (foundBook != null)
        {
            if (!foundBook.IsAvailable)
            {
                Console.WriteLine($"\n[Библиотека] Книга \"{title}\" уже выдана и недоступна.");
                return;
            }
            reader.TakeBook(foundBook);
        }
        else
        {
            Console.WriteLine($"\n[Библиотека] Книга \"{title}\" не найдена для читателя {reader.Name}.");
        }
    }
}

// ==========================================


Library myLibrary = new Library();

// ===== Простое наследование =====
myLibrary.AddBook(new Textbook("Алгебра", "Иванов И.И.", 2020, "Математика", 9, true, "MATH-9"));
myLibrary.AddBook(new Fiction("Гарри Поттер", "Дж. К. Роулинг", 1997, "Фэнтези", "12+", true, "Гарри Поттер"));
myLibrary.AddBook(new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, "Физика", "10.1000/xyz", 1500, true));

// ===== Сложное (многоуровневое) наследование =====
IllustratedTextbook illustrated = new IllustratedTextbook(
    "Геометрия в иллюстрациях", "Петров П.П.", 2021, "Математика",
    7, true, "GEO-7", "Сидоров С.С.", 120, "PDF");
myLibrary.AddBook(illustrated);

// ===== Множественное наследование через интерфейсы =====
MultimediaBook multimedia = new MultimediaBook(
    "Интерактивный курс физики", "Смирнов А.А.", 2022,
    "978-5-00000-000-0", 300, "Наука", true,
    "MP4", 90, true);
myLibrary.AddBook(multimedia);

// Создание читателей
Reader reader1 = new Reader("Алексей");
Reader reader2 = new Reader("Мария");

Console.WriteLine("\n=== Демонстрация взаимодействия объектов ===");

myLibrary.LendBook(reader1, "Гарри Поттер");
myLibrary.LendBook(reader2, "Алгебра");
myLibrary.LendBook(reader1, "Война и мир");

Console.WriteLine("\n=== Демонстрация новых возможностей ===");

Console.WriteLine(illustrated.GetInfo());
illustrated.ShowIllustration(5);

Console.WriteLine(multimedia.GetInfo());
multimedia.Download();
multimedia.Listen();
multimedia.LaunchInteractiveMode();

multimedia.ReturnBook();
myLibrary.LendBook(reader2, "Интерактивный курс физики");
```
