```
using System;
using System.Collections.Generic;

// ===== ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ =====
public interface IDigitalMedia
{
    string FileFormat { get; set; }
    void Download();
}

public interface IAudioBook
{
    int DurationMinutes { get; set; }
    void Listen();
}

public interface IInteractive
{
    void LaunchInteractiveMode();
}

// 1. Базовый класс Book
public class Book
{
    private string _title;
    private string _author;
    private int _yearOfPublication;
    private string _isbn;
    private int _pages;
    private string _publisher;
    private bool _isAvailable;

    // СПРИНТ 3: Новые атрибуты базового класса
    private string _language;
    private double _rating;
    private decimal _price;
    private bool _isDamaged;

    public string Title { get { return _title; } set { _title = value; } }
    public string Author { get { return _author; } set { _author = value; } }
    public int YearOfPublication { get { return _yearOfPublication; } set { _yearOfPublication = value; } }
    public string ISBN { get { return _isbn; } set { _isbn = value; } }
    public int Pages { get { return _pages; } set { _pages = value; } }
    public string Publisher { get { return _publisher; } set { _publisher = value; } }
    public bool IsAvailable { get { return _isAvailable; } set { _isAvailable = value; } }
    
    // СПРИНТ 3: Свойства для новых атрибутов
    public string Language { get { return _language; } set { _language = value; } }
    public double Rating { get { return _rating; } set { _rating = value; } }
    public decimal Price { get { return _price; } set { _price = value; } }
    public bool IsDamaged { get { return _isDamaged; } set { _isDamaged = value; } }

    public Book(string title, string author, int yearOfPublication,
                string isbn = "N/A", int pages = 0,
                string publisher = "Неизвестно", bool isAvailable = true,
                string language = "Русский", double rating = 5.0, decimal price = 0, bool isDamaged = false)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        ISBN = isbn;
        Pages = pages;
        Publisher = publisher;
        IsAvailable = isAvailable;
        Language = language;
        Rating = rating;
        Price = price;
        IsDamaged = isDamaged;
    }

    // СПРИНТ 3: Перекрытие (Overriding) - базовый виртуальный метод
    public virtual string GetFullDescription()
    {
        return $"[Книга] \"{Title}\" | Автор: {Author} | Язык: {Language} | Рейтинг: {Rating}";
    }

    public virtual string GetInfo()
    {
        return $"Название: \"{Title}\", Автор: {Author}, Год: {YearOfPublication}, ISBN: {ISBN}, Цена: {Price} руб.";
    }

    // СПРИНТ 3: Перегрузка (Overloading) - методы с одинаковым именем, но разными параметрами
    public decimal CalculateFine(int days)
    {
        return CalculateFine(days, 10.0m); // 10 рублей по умолчанию
    }

    public decimal CalculateFine(int days, decimal dailyRate)
    {
        if (!IsDamaged) return 0;
        return days * dailyRate;
    }

    // СПРИНТ 3: Новый метод базового класса
    public void ApplyDiscount(decimal discountPercent)
    {
        Price -= Price * (discountPercent / 100m);
        Console.WriteLine($"На книгу \"{Title}\" применена скидка. Новая цена: {Price} руб.");
    }

    public virtual void Read() { Console.WriteLine($"Чтение книги \"{Title}\"..."); }
    public virtual void Borrow()
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"Книга \"{Title}\" недоступна.");
            return;
        }
        IsAvailable = false;
        Console.WriteLine($"Книга \"{Title}\" выдана.");
    }
}

// 2. Textbook
public class Textbook : Book
{
    private string _subject;
    
    // СПРИНТ 3: Новые атрибуты
    private string _curriculumStandard;
    private int _gradeLevel;

    public string Subject { get { return _subject; } set { _subject = value; } }
    public string CurriculumStandard { get { return _curriculumStandard; } set { _curriculumStandard = value; } }
    public int GradeLevel { get { return _gradeLevel; } set { _gradeLevel = value; } }

    public Textbook(string title, string author, int year, string subject, int gradeLevel, string curriculum)
        : base(title, author, year)
    {
        Subject = subject;
        GradeLevel = gradeLevel;
        CurriculumStandard = curriculum;
    }

    // СПРИНТ 3: Перекрытие (Overriding)
    public override string GetFullDescription()
    {
        return $"[Учебник] \"{Title}\" | Предмет: {Subject} | Класс: {GradeLevel} | Стандарт: {CurriculumStandard}";
    }

    // СПРИНТ 3: Новые методы
    public void AssignHomework(int exerciseCount)
    {
        Console.WriteLine($"По учебнику \"{Title}\" задано {exerciseCount} упражнений.");
    }

    public override void Read() { Console.WriteLine($"Чтение учебника \"{Title}\" по предмету: {Subject}."); }
}

// 3. Fiction
public class Fiction : Book
{
    private string _genre;
    
    // СПРИНТ 3: Новые атрибуты
    private string _mainCharacter;
    private bool _hasFilmAdaptation;

    public string Genre { get { return _genre; } set { _genre = value; } }
    public string MainCharacter { get { return _mainCharacter; } set { _mainCharacter = value; } }
    public bool HasFilmAdaptation { get { return _hasFilmAdaptation; } set { _hasFilmAdaptation = value; } }

    public Fiction(string title, string author, int year, string genre, string mainCharacter, bool hasFilm)
        : base(title, author, year)
    {
        Genre = genre;
        MainCharacter = mainCharacter;
        HasFilmAdaptation = hasFilm;
    }

    // СПРИНТ 3: Перекрытие (Overriding)
    public override string GetFullDescription()
    {
        return $"[Худ. литература] \"{Title}\" | Жанр: {Genre} | Главный герой: {MainCharacter} | Экранизация: {(HasFilmAdaptation ? "Да" : "Нет")}";
    }

    // СПРИНТ 3: Новые методы
    public void ReadChapter(int chapter)
    {
        Console.WriteLine($"Чтение {chapter}-й главы книги \"{Title}\".");
    }

    public override void Borrow()
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"Художественная книга \"{Title}\" недоступна.");
            return;
        }
        IsAvailable = false;
        Console.WriteLine($"Выдача художественной книги \"{Title}\" в жанре: {Genre}.");
    }
}

// 4. ScientificLiterature
public class ScientificLiterature : Book
{
    private string _fieldOfScience;
    
    // СПРИНТ 3: Новые атрибуты
    private string _doi;
    private int _impactFactor;

    public string FieldOfScience { get { return _fieldOfScience; } set { _fieldOfScience = value; } }
    public string DOI { get { return _doi; } set { _doi = value; } }
    public int ImpactFactor { get { return _impactFactor; } set { _impactFactor = value; } }

    public ScientificLiterature(string title, string author, int year, string fieldOfScience, string doi, int impactFactor)
        : base(title, author, year)
    {
        FieldOfScience = fieldOfScience;
        DOI = doi;
        ImpactFactor = impactFactor;
    }

    // СПРИНТ 3: Перекрытие (Overriding)
    public override string GetFullDescription()
    {
        return $"[Научная литература] \"{Title}\" | Область: {FieldOfScience} | DOI: {DOI} | Импакт-фактор: {ImpactFactor}";
    }

    // СПРИНТ 3: Новые методы
    public void AddCitation(int count)
    {
        Console.WriteLine($"Статья \"{Title}\" процитирована {count} раз(а).");
    }
}

// ===== СЛОЖНОЕ НАСЛЕДОВАНИЕ =====
public class IllustratedTextbook : Textbook
{
    private string _illustrator;
    public string Illustrator { get { return _illustrator; } set { _illustrator = value; } }

    public IllustratedTextbook(string title, string author, int year, string subject, int gradeLevel, string curriculum, string illustrator)
        : base(title, author, year, subject, gradeLevel, curriculum)
    {
        Illustrator = illustrator;
    }

    public override string GetFullDescription()
    {
        return base.GetFullDescription() + $" | Иллюстратор: {Illustrator}";
    }
}

// ===== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ =====
public class MultimediaBook : Book, IDigitalMedia, IAudioBook, IInteractive
{
    private string _fileFormat;
    private int _durationMinutes;

    public string FileFormat { get { return _fileFormat; } set { _fileFormat = value; } }
    public int DurationMinutes { get { return _durationMinutes; } set { _durationMinutes = value; } }

    public MultimediaBook(string title, string author, int year, string fileFormat, int durationMinutes)
        : base(title, author, year)
    {
        FileFormat = fileFormat;
        DurationMinutes = durationMinutes;
    }

    public void Download() { Console.WriteLine($"Скачивание книги \"{Title}\" в формате {FileFormat}."); }
    public void Listen() { Console.WriteLine($"Прослушивание аудиоверсии \"{Title}\" ({DurationMinutes} мин.)."); }
    public void LaunchInteractiveMode() { Console.WriteLine($"Запущен интерактивный режим книги \"{Title}\"."); }

    public override string GetFullDescription()
    {
        return $"[Мультимедиа] \"{Title}\" | Формат: {FileFormat} | Длительность: {DurationMinutes} мин.";
    }
}

// СПРИНТ 3: Generic-класс (обобщенный класс)
// Позволяет хранить книги только определенного типа
public class LibraryCatalog<T> where T : Book
{
    private List<T> _items = new List<T>();

    public void AddItem(T item)
    {
        _items.Add(item);
        Console.WriteLine($"[Каталог] Добавлен элемент: {item.Title}");
    }

    public void ShowAll()
    {
        Console.WriteLine($"\n--- Содержимое каталога ({typeof(T).Name}) ---");
        foreach (var item in _items)
        {
            Console.WriteLine(item.GetFullDescription()); // Вызов полиморфного метода
        }
    }
}

// 5. Читатель
public class Reader
{
    public string Name { get; set; }
    public Reader(string name) { Name = name; }

    // СПРИНТ 3: Перегрузка метода TakeBook (Overloading)
    public void TakeBook(Book book)
    {
        Console.WriteLine($"\n[Читатель {Name}] берет книгу \"{book.Title}\".");
        book.Borrow();
    }

    public void TakeBook(List<Book> books)
    {
        Console.WriteLine($"\n[Читатель {Name}] берет сразу {books.Count} книг(и).");
        foreach (var b in books) { b.Borrow(); }
    }
}

// ===== ВЫПОЛНЕНИЕ ПРОГРАММЫ (БЕЗ class Program ДЛЯ JUPYTER) =====

Console.WriteLine("=== ДЕМОНСТРАЦИЯ СПРИНТ 3 ===");

// 1. Демонстрация Generic-класса
LibraryCatalog<Textbook> textbookCatalog = new LibraryCatalog<Textbook>();
var t1 = new Textbook("Алгебра", "Иванов", 2020, "Математика", 9, "ФГОС");
textbookCatalog.AddItem(t1);
textbookCatalog.ShowAll();

LibraryCatalog<Fiction> fictionCatalog = new LibraryCatalog<Fiction>();
var f1 = new Fiction("Гарри Поттер", "Роулинг", 1997, "Фэнтези", "Гарри", true);
fictionCatalog.AddItem(f1);
fictionCatalog.ShowAll();

// 2. Демонстрация Полиморфизма (Перекрытие)
Console.WriteLine("\n--- Демонстрация перекрытия методов (Overriding) ---");
Book myBook = new Fiction("Властелин Колец", "Толкин", 1954, "Фэнтези", "Фродо", true);
Console.WriteLine(myBook.GetFullDescription()); // Вызовется метод из Fiction, хотя переменная типа Book

// 3. Демонстрация Перегрузки (Overloading)
Console.WriteLine("\n--- Демонстрация перегрузки методов (Overloading) ---");
myBook.IsDamaged = true;
decimal fine1 = myBook.CalculateFine(5);          // Вызов первой версии
decimal fine2 = myBook.CalculateFine(5, 15.5m);   // Вызов второй версии
Console.WriteLine($"Штраф (по умолчанию): {fine1} руб.");
Console.WriteLine($"Штраф (свой тариф): {fine2} руб.");

// 4. Демонстрация перегрузки метода у Читателя
Console.WriteLine("\n--- Демонстрация перегрузки TakeBook ---");
Reader reader = new Reader("Алексей");
reader.TakeBook(myBook);
reader.TakeBook(new List<Book> { t1, f1 });
```
