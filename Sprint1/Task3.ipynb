<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).
• Методы:
o GetInfo(): метод для получения информации о книге в виде строки.
o Read(): метод для вывода сообщения о чтении книги.
o Borrow(): метод для вывода сообщения о выдаче книги на чтение.
Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// 1. Базовый класс Book
public class Book
{
    // Приватные поля (для инкапсуляции)
    private string _title;
    private string _author;
    private int _yearOfPublication;

    // Публичные свойства (Геттеры и Сеттеры)
    public string Title
    {
        get { return _title; }
        set { _title = value; }
    }

    public string Author
    {
        get { return _author; }
        set { _author = value; }
    }

    public int YearOfPublication
    {
        get { return _yearOfPublication; }
        set { _yearOfPublication = value; }
    }

    // Конструктор использует сеттеры (через свойства)
    public Book(string title, string author, int yearOfPublication)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
    }

    public virtual string GetInfo()
    {
        return $"Название: \"{Title}\", Автор: {Author}, Год издания: {YearOfPublication}";
    }

    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги \"{Title}\"...");
    }

    public virtual void Borrow()
    {
        Console.WriteLine($"Книга \"{Title}\" выдана на чтение.");
    }
}

// 2. Производный класс Учебник (Textbook)
public class Textbook : Book
{
    private string _subject;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public Textbook(string title, string author, int year, string subject) 
        : base(title, author, year)
    {
        Subject = subject;
    }

    public override void Read()
    {
        Console.WriteLine($"Чтение учебника \"{Title}\" по предмету: {Subject}.");
    }
}

// 3. Производный класс Художественная литература (Fiction)
public class Fiction : Book
{
    private string _genre;

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public Fiction(string title, string author, int year, string genre) 
        : base(title, author, year)
    {
        Genre = genre;
    }

    public override void Borrow()
    {
        Console.WriteLine($"Выдача художественной книги \"{Title}\" в жанре: {Genre}.");
    }
}

// 4. Производный класс Научная литература (ScientificLiterature)
public class ScientificLiterature : Book
{
    private string _fieldOfScience;

    public string FieldOfScience
    {
        get { return _fieldOfScience; }
        set { _fieldOfScience = value; }
    }

    public ScientificLiterature(string title, string author, int year, string fieldOfScience) 
        : base(title, author, year)
    {
        FieldOfScience = fieldOfScience;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {FieldOfScience}";
    }
}

// 5. Класс Читатель (Reader) - участник взаимодействия
public class Reader
{
    public string Name { get; set; }

    public Reader(string name)
    {
        Name = name;
    }

    // Метод взаимодействия с книгой
    public void TakeBook(Book book)
    {
        Console.WriteLine($"\n[Читатель {Name}] берет книгу \"{book.Title}\".");
        book.Borrow(); // Вызов полиморфного метода
    }
}

// 6. Класс Библиотека (Library) - управляет книгами и взаимодействием
public class Library
{
    private List<Book> _books = new List<Book>();

    public void AddBook(Book book)
    {
        _books.Add(book);
        Console.WriteLine($"[Библиотека] Добавлена книга: \"{book.Title}\"");
    }

    // Метод поиска и выдачи книги читателю
    public void LendBook(Reader reader, string title)
    {
        Book foundBook = null;
        
        // Поиск книги в библиотеке
        foreach (var book in _books)
        {
            if (book.Title.Equals(title, StringComparison.OrdinalIgnoreCase))
            {
                foundBook = book;
                break;
            }
        }

        if (foundBook != null)
        {
            // Взаимодействие объектов: Библиотека передает книгу Читателю
            reader.TakeBook(foundBook);
        }
        else
        {
            Console.WriteLine($"\n[Библиотека] Книга \"{title}\" не найдена для читателя {reader.Name}.");
        }
    }
}

class Program
{
    static void Main(string[] args)
    {
        // Инициализация библиотеки
        Library myLibrary = new Library();

        // Создание книг и добавление их в библиотеку
        myLibrary.AddBook(new Textbook("Алгебра", "Иванов И.И.", 2020, "Математика"));
        myLibrary.AddBook(new Fiction("Гарри Поттер", "Дж. К. Роулинг", 1997, "Фэнтези"));
        myLibrary.AddBook(new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, "Физика"));

        // Создание читателей
        Reader reader1 = new Reader("Алексей");
        Reader reader2 = new Reader("Мария");

        Console.WriteLine("\n=== Демонстрация взаимодействия объектов ===");

        // Взаимодействие: Читатель просит библиотеку выдать книгу
        myLibrary.LendBook(reader1, "Гарри Поттер");
        myLibrary.LendBook(reader2, "Алгебра");
        myLibrary.LendBook(reader1, "Война и мир"); // Книги нет в библиотеке

        Console.ReadLine();
    }
}
```
